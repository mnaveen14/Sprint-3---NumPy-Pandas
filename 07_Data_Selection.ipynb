# **Notebook 7 – Selecting & Filtering Data**

## **1. loc**



**Understand the Concept**

* loc is used to select rows and columns using labels.
* We can select a specific row, column, or a group of rows and columns.
* It is useful when we know the row labels or column names.




**Demonstrate the Concept**

**Real-Life Example:**

* A company wants to view specific employee information.
* loc can select the required rows and columns using their labels.

**AI/ML Usage:**

* loc is used to select specific rows and features during data preparation.

```python
# Import Libraries
import pandas as pd
```

```python
# Load Dataset
employees = pd.read_csv("employees_dataset.csv")
```

```python
# Implement the Concept:
# loc - select specific row & col
employees.loc[10:15:2, ["First Name", "Salary"]] # loc[start:stop:step]
```

**Interpretation**

* loc selects rows using their index labels.
* It also selects the required columns using their column names.


## **2. iloc**



**Understand the Concept**

* iloc is used to select rows and columns using their position.
* The position starts from 0.
* We can select a specific row, column, or a group of rows and columns.



**Demonstrate the Concept**

**Real-Life Example:**

* A company wants to view a particular set of employee records.
* iloc can select the required records based on their position in the dataset.

**AI/ML Usage:**

* iloc is used to select specific rows and columns during data preparation.

```python
# Implement the Concept:
# iloc - select row & col based on position(index)
employees.iloc[5:7, 0:4] # iloc[row,col] , end - n-1
```

iloc selects data using row and column positions

## **3. Boolean Indexing**



**Understand the Concept**

* Boolean Indexing is used to select rows based on a condition.
* It returns True for rows that match the condition and False for other rows.
* We can use conditions such as greater than, less than, or equal to.
* It is useful for filtering specific data.



**Demonstrate the Concept**

**Real-Life Example:**

* A company wants to find employees with higher salaries.
* Boolean Indexing can select the employees who meet the salary condition.

**AI/ML Usage:**

* Boolean Indexing is used to filter data based on feature conditions during data preparation.

```python
# Implement the Concept:
employees[employees["Salary"] > 145000].head()
```

Its show only salary greater than 145000 and 1st 5 rows

## **4. Filtering Rows**



**Understand the Concept**

* Filtering Rows means selecting only the rows that match a condition.
* We can use conditions to find specific records from a DataFrame.
* Multiple conditions can also be used together.



**Demonstrate the Concept**


**Real-Life Example:**

* A company wants to view employees from a particular team.
* Filtering Rows helps display only the required employee records.

**AI/ML Usage:**

* Filtering Rows is used to select relevant records before data analysis and Machine Learning.

```python
# Implement the Concept:
# Filter  - employees from engineering team
engineering_employees = employees[employees["Team"] == "Engineering"]

print(engineering_employees.sample(3))
```

**Interpretation**

* The condition checks the Team column.
* Only employees belonging to the Engineering team are selected.

## **5. query()**



**Understand the Concept**

* query() is used to filter rows using a condition written as a string.
* It makes filtering easier to read and write.
* We can use column names directly inside the condition.




**Demonstrate the Concept**

**Real-Life Example:**

* A company wants to find employees with a salary below a certain level.
* query() can quickly filter the required employee records.

**AI/ML Usage:**

* query() is used to select relevant records during data preparation.

```python
# Implement the Concept:
# query() - Filter salary < 50000
senior = employees.query("Salary < 50000")

print(senior)
```

query() provides a simple way to write filtering conditions

## **6. isin()**



**Understand the Concept**

* isin() is used to filter rows that contain values from a given list.
* It is useful when we want to check multiple values in a column.
* It returns rows where the column value matches any value in the list.




**Demonstrate the Concept**

**Real-Life Example:**

* A company wants to view employees from a few selected teams.
* isin() helps filter those teams at the same time.

**AI/ML Usage:**

* isin() is used to select specific categories or groups during data preparation.

```python
# isin() - Select employees from selected teams
selected_teams = employees[employees["Team"].isin(["Finance", "IT"])]

print(selected_teams)
```

**Interpretation**

* isin() checks whether each Team value is present in the given list.
* Only employees from  Finance, and IT are selected.

## **7. between()**



**Understand the Concept**

* between() is used to filter rows where values fall within a specified range.
* It works mainly with numerical or date columns.
* By default, the start and end values are included.



**Demonstrate the Concept**

**Real-Life Example:**

* A company wants to find employees whose salaries fall within a particular salary range.
* between() helps select only those employee records.

**AI/ML Usage:**

* between() is used to filter numerical features within a specific range during data preparation.

```python
# Implement the Concept:
# between() - salary between 70000 and 100000
salary_range = employees[employees["Salary"].between(70000, 100000)]
print(salary_range)
```

between() selects employees whose salary is within the specified range.

## **8. Sorting Data**

**Understand the Concept**

* Sorting Data is used to arrange rows based on the values in a column.
* sort_values() is used to sort a DataFrame.
* Data can be sorted in ascending or descending order.
* We can sort using one or multiple columns.



**Demonstrate the Concept**

**Real-Life Example:**

* A company wants to view employees from the highest salary to the lowest salary.
* Sorting Data helps arrange the employee records in the required order.

**AI/ML Usage:**

* Sorting Data helps organize and inspect features during data analysis and preprocessing.

```python
# sort_values() - salary in descending order
sorted_data = employees.sort_values(by="Salary", ascending=False)

print(sorted_data)
```

**Interpretation**

* sort_values() arranges the employee records based on Salary.
* ascending=False sorts the salary from highest to lowest.
* We can use ascending=True to sort from lowest to highest.

## **Best Practices**

* Use loc when selecting data by labels.
* Use iloc when selecting data by position.
* Use clear conditions when filtering rows.
* Use isin() when filtering multiple values.
* Use between() when filtering values within a range.
* Use sort_values() to arrange data clearly.

## **Common Mistakes**

* Confusing loc and iloc.
* Using incorrect column names.
* Forgetting that iloc uses zero-based positions.
* Using incorrect conditions while filtering.
* Forgetting ascending=False when descending order is required.

## **Advantages**

* Makes data selection easy.
* Helps filter only the required records.
* Supports selection using labels and positions.
* Makes data analysis more efficient.

## **Limitations**

* Incorrect conditions can return unexpected results.
* Large filtering operations may require more memory.
* Complex conditions can become difficult to read.
