## **1. Vectorization**

**Understand the Concept**

* Vectorization means doing an operation on a whole column at once.
* We do not need to use a loop for each row.

**Demonstrate the Concept**

**Real-Life Example:**

* An online shopping company needs to calculate the total amount for thousands of transactions.
* Vectorization can calculate Quantity × Price for the entire dataset without using a loop.

**AI/ML Usage:**

* Vectorization is useful for fast feature creation and data preprocessing on large datasets.

```python
# Implement the Concept:
# Vectorization - Quantity × Price = Total_Amount 
shopping["Total_Amount"] = ( shopping["Quantity"] * shopping["Price"])
print(shopping[["Quantity", "Price", "Total_Amount"]].sample(7))
```

**Interpretation**

* Quantity and Price are multiplied for all rows at once.
* No loop is required.

## **2. apply() vs Loops**



**Understand the Concept**

* Both apply() and loops can perform operations on data.
* A loop processes data one row at a time.
* apply() applies a function to a column or rows.



**Demonstrate the Concept**

**Real-Life Example:**

* An online shopping company wants to group orders as High Value or Normal.
* This can be done using a loop or apply().

**AI/ML Usage:**

* apply() can be used to create new features during data preprocessing.
* For simple column calculations, vectorization is usually faster than both apply() and loops.

```python
# Implement the Concept:
# Using a loop
order_category = []

for amount in shopping["Total_Amount"]:
    if amount > 100000:
        order_category.append("High Value")
    else:
        order_category.append("Normal")

shopping["Order_Cat_Loop"] = order_category
shopping["Order_Cat_Loop"].head()
```

```python
# Using apply()
shopping["Order_Cat_Apply"] = shopping["Total_Amount"].apply(lambda amount: "High Value" if amount > 10000 else "Normal")

print(shopping[[ "Total_Amount","Order_Cat_Loop", "Order_Cat_Loop"]].head())
```

**Interpretation**

* The loop checks each order one by one.
* apply() does the same task with shorter code.
* Both methods give the same result.

## **3. Memory Optimization**



**Understand the Concept**

* Memory optimization means reducing the amount of memory used by a dataset.
* Large datasets can use a lot of memory.
* Choosing suitable data types can reduce memory usage.



**Demonstrate the Concept**

**Example:**

* A column with small whole numbers may use int64 by default.
* Changing it to int32 can use less memory.

**Real-Life Example:**

* An online shopping company stores millions of transactions.
* Reducing memory usage helps process the dataset more efficiently.

**AI/ML Usage:**

* Memory optimization helps when working with large datasets.
* It can make data loading and processing easier.



```python
# Before memory optimization
print("Before optimization:")
print(shopping[["Quantity", "Discount","Price"]].memory_usage(deep=True))
```

```python
# Implement the Concept:
# Memory Optimization - for numerical col data types
shopping["Quantity"] = shopping["Quantity"].astype("int8")
shopping["Discount"] = shopping["Discount"].astype("int8")
shopping["Price"] = shopping["Price"].astype("float32")
```

```python
# After memory optimization
print("After optimization:")
print(shopping[["Quantity", "Discount","Price"]].memory_usage(deep=True))
```

**Interpretation**

* Smaller data types can reduce memory usage.
* Quantity and Discount are changed to int8.
* Price is changed to float32.
* The dataset can use less memory after optimization.

## **4. Category Data Type**



**Understand the Concept**

* Category data type is useful for columns with a small number of repeated values.
* It can reduce memory usage.

**Demonstrate the Concept**

**Example:**

* Order_Status has only a few repeated values such as Delivered, Pending, Cancelled, and Shipped.
* Instead of storing the full text repeatedly, Pandas can store them as categories.

**Real-Life Example:**

* An online shopping dataset may have thousands of orders but only a few payment methods and order statuses.
* Converting these repeated values to category can save memory.

**AI/ML Usage:**

* Category data can help reduce memory usage when working with large datasets.
* It can also help prepare categorical data before further preprocessing.

```python
# Check data type before conversion
print(shopping["Order_Status"].dtype)
```

```python
# Category Data Type - convert Order_Status to category
shopping["Order_Status"] = shopping["Order_Status"].astype("category")
```

```python
# Check data type after conversion
print(shopping["Order_Status"].dtype)
```

**Interpretation**

* Converting object to category can use less memory.
* This is useful when a column has only a few unique values.

## **5. Efficient Pandas Practices**



**Understand the Concept**

* Efficient Pandas practices help process large datasets faster.
* They can also reduce memory usage.
* Choosing the right method depends on the dataset and task.



### **5.1 Select Only Required Columns**

Instead of working with all columns, select only the columns needed.

**Real-Life Example:**

* A company has a large shopping dataset but only needs customer, product, and price details for analysis.

```python
# Select only required columns
selected = shopping[["Customer_ID","Product_Category","Quantity","Price","Order_Status"]]
print(selected.head())
```

### **5.2 Specify Data Types**

Suitable data types can reduce memory usage.

**Real-Life Example:**

* Columns with repeated values, such as Order_Status, can be stored as category.

```python
# specified data types
shopping_data = pd.read_csv("online_shopping_dataset.csv",
                            dtype={"Quantity": "int8",
                                   "Discount": "int8",
                 "Order_Status": "category"})

print(shopping_data.dtypes)
```

### **5.3 Use query()**

query() provides a simple way to filter data.

**Real-Life Example:**

* A company wants to find orders where the product price is greater than 10000.

```python
# query() - find products with Price greater than 10000
result = shopping.query("Price > 10000")
print(result["Price"].sample(3))
```

### **5.4 Use Appropriate File Formats**

* CSV is useful for sharing data.
* Parquet can be useful for storing and processing large datasets.

**Real-Life Example:**

* A company repeatedly processes a large transaction dataset.
* Saving it in Parquet format can make repeated data processing more efficient.

```python
# Appropriate File Formats - Save dataset as a Parquet file
shopping.to_parquet("online_shopping_dataset.parquet",index=False)

shopping_parquet = pd.read_parquet("online_shopping_dataset.parquet")
print(shopping_parquet.head())
```

### **5.5 Read Large Files in Chunks**

* Large CSV files can be read in smaller parts using chunksize.
* This helps when the full dataset is too large to load into memory.

**Real-Life Example:**

* A company has millions of shopping transactions.
* Reading the data in chunks allows processing without loading everything at once.

```python
# Read the dataset in chunks
# chunksize - no of rows = row divide by chunksize
for chunk in pd.read_csv("online_shopping_dataset.csv",chunksize=2000): 
    print(chunk.shape)
```

## **Best Practices**

* Use vectorized operations instead of loops when possible.
* Select only the columns you need.
* Use suitable data types to reduce memory usage.
* Convert repeated text values to category when appropriate.
* Use chunks when working with very large CSV files.
* Use efficient file formats when needed.

## **Common Mistakes**

* Using loops for simple column calculations.
* Loading unnecessary columns.
* Using large data types when smaller types are enough.
* Converting columns to category without checking their values.
* Loading a very large file fully into memory when chunks can be used.

## **Advantages**

* Makes data processing faster.
* Reduces memory usage.
* Helps handle large datasets.
* Improves Pandas performance.
* Makes data preprocessing more efficient.

## **Limitations**

* Optimization is not always needed for small datasets.
* Incorrect data type conversion can cause data loss or errors.
* apply() and loops may still be needed for complex operations.
* Chunk processing can make the code more complex.
