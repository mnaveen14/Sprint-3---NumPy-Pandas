# **Notebook 10 – Combining Datasets**

## **1. concat()**



**Understand the Concept**

* concat() is used to combine two or more DataFrames.
* It can combine DataFrames vertically (rows) or horizontally (columns).
* By default, it combines DataFrames row-wise.



**Demonstrate the Concept**

**Example:**

* One dataset contains existing customers.
* Another dataset contains new customers.
* concat() combines both datasets into one DataFrame.


**AI/ML Usage:**

* `concat()` is used to combine datasets from multiple sources before data preprocessing and model training.

```python
# Libraries: 
import pandas as pd
```

```python
# Load Dataset: 
customers = pd.read_csv("customers_dataset.csv")
customers.head(3)
```

```python
# Implement the Concept:
# New customers data 
new_customers = pd.DataFrame({"Customer_ID" : ["CUST0501", "CUST0502"],
                              "first_name": ["Arun", "Kiran"],
                              "last_name": ["Kumar", "Das"],
                              "City": ["Chennai", "Pune"],
                              "mail": ["arun.kumar501@example.com", "kiran.das502@example.com"]})
new_customers
```

```python
# Combine both DataFrames
combined_customers = pd.concat([customers,new_customers], ignore_index = True)
combined_customers.tail()
```

**Interpretation**

* concat() combines the existing customer data with the new customer records.
* The new rows are added below the existing 500 rows.
* The final DataFrame contains 502 rows.
* ignore_index=True creates a new continuous index.

## **2. merge()**



**Understand the Concept**

* merge() is used to combine two DataFrames based on a common column or key.
* It works similarly to SQL joins.
* Common merge types:
    * inner join - Returns only matching records from both DataFrames.
    * left join - Returns all records from the left dataset and matching records from the right.
    * right join - Returns all records from the right dataset and matching records from the left.
    * outer join - Returns all records from both dataset, including non-matching records.



**Demonstrate the Concept**

**Example:**

* One dataset contains customer information.
* Another dataset contains order information.
* Both datasets contain Customer_ID.
* merge() combines the matching records using Customer_ID.

**Real-Life Example:**

* A company stores customer details and order details in separate datasets.
* merge() can combine both datasets for customer order analysis.

**AI/ML Usage:**

* merge() is used to combine features from different datasets before data preprocessing and model training.

```python
# orders data
orders = pd.DataFrame({"Order_ID": ["ORD001", "ORD002", "ORD003", "ORD004", "ORD005"],
                       "Customer_ID": ["CUST0001", "CUST0002", "CUST0003", "CUST0004", "CUST0005"],
                       "Product": ["Laptop", "Mobile", "Headphones", "Keyboard", "Mouse"],
                       "Order_Value": [65000, 30000, 2500, 3500, 1200]})
orders
```

```python
# Implement the Concept:
# Merge - customers & orders dataset 
customer_orders = pd.merge(customers, orders, on="Customer_ID", how="inner")
customer_orders
```

**Interpretation**

* merge() combines the customers and orders dataset.
* Customer_ID is used as the common column.
* Only records with matching Customer_ID values are included because how="inner" is used.
* The result contains customer details along with their order information.

## **3. join()**



**Understand the Concept**

* join() is used to combine two dataset based on their index by default.
* It can also combine dataset using a common column.



**Demonstrate the Concept**

**Real-Life Example:**

* A company stores customer details and membership details separately.
* join() can combine both datasets into a single DataFrame.

**AI/ML Usage:**

* join() can be used to combine related features from multiple datasets before data preprocessing and model training.

```python
# membership ddta
membership = pd.DataFrame({"Customer_ID" :["CUST0001","CUST0002","CUST0003","CUST0004","CUST0005"],
                           "Membership": ["Gold", "Silver", "Gold", "Platinum", "Silver"]})
membership
```

```python
# Implement the Concept:
# join() - join customers & membership using common col 
customer_membership = customers.join(membership.set_index("Customer_ID"), on = "Customer_ID")
customer_membership[2:8]
```

**Interpretation**

* join() combines the customers and membership data using Customer_ID.
* Customer_ID acts as the common column between both data.
* The Membership information is added to the matching customer records.
* Customers without matching membership information will contain missing values.

**Demonstrate the Concept**

**Real-Life Example:**

* pd.concat() combines the existing and new records.

**AI/ML Usage:**

* pd.concat() is used to combine new data with existing datasets during data collection and preprocessing.

```python
# new customers data 
new_customers # already define
```

```python
# Implement the Concept: 
# concat() - customers & new_customers
updated_customers = pd.concat([customers, new_customers],ignore_index=True)
print(updated_customers.tail())
```

**Interpretation**

* DataFrame.append() is deprecated and should not be used in newer Pandas versions.

## **5. combine_first()**



**Understand the Concept**

* combine_first() is used to combine two DataFrames and fills missing values in one DataFrame using values from another DataFrame.
* Existing non-missing values in the first DataFrame are kept unchanged.



**Demonstrate the Concept**

**Example:**

* One customer dataset contains missing City and mail values.
* Another dataset contains the missing information.
* combine_first() fills the missing values using the second DataFrame.

**Real-Life Example:**

* A company has incomplete customer records in one dataset and updated information in another dataset.
* combine_first() can fill the missing details from the updated dataset.

**AI/ML Usage:**

* combine_first() can be used to fill missing values by combining information from multiple datasets during data preprocessing.

```python
# Implement the Concept: 
# customer data with missing values
customers_1 = pd.DataFrame({"Customer_ID": ["CUST0001", "CUST0002", "CUST0003","CUST0004"],
                            "City": ["Chennai", None, "Mumbai","Coimbatore"],
                            "mail": ["arjun@example.com", None, "rahul@example.com",None]})
customers_1
```

```python
# customer data with non-missing values
customers_2 = pd.DataFrame({"Customer_ID": ["CUST0001", "CUST0002", "CUST0003","CUST0005"],
                            "City": ["Chennai", "Bengaluru", "Mumbai","Dehli"],
                            "mail": ["arjun@example.com","vikram@example.com","rahul@example.com","navin@emanple.com"]})
customers_2
```

```python
# index - both data customer_id is a index 
customers_1 = customers_1.set_index("Customer_ID")
customers_2 = customers_2.set_index("Customer_ID")
```

```python
# combine_first - fill missing values using 2nd data
combined_customers = customers_1.combine_first(customers_2)
print(combined_customers)
```

**Interpretation**

* combine_first() keeps the existing values from customers_1.
* Missing values in customers_1 are filled using matching values from customers_2.
* For CUST0002, the missing City and mail values are filled from customers_2.
* CUST0004 exists only in customers_1, so its missing mail value remains NaN.
* CUST0005 exists only in customers_2, so it is added to the combined DataFrame.

## **Best Practices**

* Use concat() to combine DataFrames row-wise or column-wise.
* Use merge() when DataFrames have a common column or key.
* Use join() when combining DataFrames based on an index or common column.
* Use pd.concat() instead of the deprecated append() method.
* Use combine_first() to fill missing values from another DataFrame.
* Check the result after combining datasets to identify missing or duplicate records.

## **Common Mistakes**

* Using the wrong common column while merging DataFrames.
* Forgetting to check for duplicate keys before merge().
* Confusing concat() with merge().
* Using append() in newer Pandas versions.
* Not checking missing values after combining datasets.
* Using the wrong join type.

## **Advantages**

* Combines information from multiple datasets.
* Helps create complete datasets for analysis.
* Supports different types of joins and combinations.
* Useful for data integration and preprocessing.
* Helps prepare data from multiple sources for AI/ML.

## **Limitations**

* Incorrect keys can produce incorrect results.
* Duplicate keys can create unexpected repeated rows.
* Large merge operations can require more memory.
* Missing values may appear when records do not match.

**7. What happens when merge() finds no matching values?**
