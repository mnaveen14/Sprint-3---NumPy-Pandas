## **1. Rename Columns**

**Understand the Concept**

* rename() is used to change the names of DataFrame columns.
* We can rename one column or multiple columns.

**Demonstrate the Concept**

**Real-Life Example:**

* A company receives employee data with unclear column names.
* Renaming columns makes the dataset easier to understand and work with.

**AI/ML Usage:**

* Clear column names make feature selection and data preprocessing easier in AI/ML.

```python
# Import Libraries
import pandas as pd
```

```python
# Load Dataset
employees = pd.read_csv("employees_dataset.csv")
```

```python
# Implement the Concept:
# Rename col 
employees = employees.rename(columns={"First Name": "Employee Name", "Bonus %": "Bonus"})
print(employees.head())
```

## **2. Drop Rows**

**Understand the Concept**

* drop() is used to remove rows from a DataFrame.
* We can remove one row or multiple rows using their index labels.



**Demonstrate the Concept**

**Real-Life Example:**

* A company finds some employee records that should not be included in the analysis.
* Drop Rows can remove those unwanted records.

**AI/ML Usage:**

* Drop Rows is used to remove unwanted records before data preprocessing and model training.

```python
# Implement the Concept:
# drop() - for row using index values
employees = employees.drop(index=[3,4])
print(employees.head())
```

**Interpretation**

* drop() removes the rows with index 0,1,3,4.
* The remaining rows are kept in the DataFrame.

## **3. Drop Columns**



**Understand the Concept**

* drop() is used to remove one or more columns from a DataFrame.
* We use axis=1 to specify that we want to remove columns.




**Demonstrate the Concept**

**Real-Life Example:**

* A company has employee information that contains unnecessary details.
* Drop Columns can remove information that is not needed for a particular analysis.

**AI/ML Usage:**

* Drop Columns is used to remove irrelevant features before Machine Learning.

```python
# Implement the Concept:
# drop() - for col use col name or axis
employees = employees.drop(columns=["Gender"])
print(employees.head())
```

**Interpretation**

* drop() removes the Gender column from the DataFrame.
* The remaining columns are kept unchanged.

## **4. Insert Columns**



**Understand the Concept**

* insert() is used to add a new column at a specific position in a DataFrame.
* We can specify the position, column name, and values for the new column.



**Demonstrate the Concept**

**Real-Life Example:**

* A company wants to add calculated employee information to its existing dataset.
* insert() can place the new information at the required position.

**AI/ML Usage:**

* insert() can be used to add newly created features to a dataset during feature engineering.

```python
# Implement the Concept:
# insert() - add new col "Annual Salary"
employees.insert(3,"Annual Salary", employees["Salary"] * 12) # position, col_name, value
print(employees.head())
```

insert() adds the Annual Salary column at position 3

## **5. assign()**



**Understand the Concept**

* assign() is used to create new columns or modify existing columns in a DataFrame.
* It returns a new DataFrame without directly changing the original DataFrame.


**Demonstrate the Concept**

**Real-Life Example:**

* A company wants to calculate additional employee information from existing data.
* assign() can create the required calculated columns.

**AI/ML Usage:**

* assign() is useful for creating new features during data preprocessing and feature engineering.

```python
# Implement the Concept:
# assign() - create new col
updated_employees = employees.assign(Annual_Salary=employees["Salary"] * 12)
print(updated_employees.head())
```

```python
# assign() - Modify the Salary column
updated_employees = employees.assign(Salary=employees["Salary"] + 5000)
print(updated_employees.head())
```

assign() returns a new DataFrame; the original employees DataFrame is not changed

## **6. apply()**



**Understand the Concept**

* apply() is used to apply a function to values in a Series or DataFrame.
* It is useful when we need to perform a custom operation on data.
* We can use built-in functions or our own functions with apply().



**Demonstrate the Concept**

**Real-Life Example:**

* A company wants to calculate an updated salary for every employee.
* apply() can apply the same calculation to each salary value.

**AI/ML Usage:**

* apply() is used for custom data transformations during data preprocessing and feature engineering.

```python
# Implement the Concept:
# apply() - add 5k in salary
employees["Updated Salary"] = employees["Salary"].apply(lambda salary: salary + 5000)
print(employees.head())
```

**Interpretation**

* apply() applies the given function to each value in the Salary column.
* 5000 is added to every employee's salary.
* The result is stored in a new Updated Salary column.

## **7. map()**



**Understand the Concept**

* map() is used to apply a function or mapping to each value in a Series.
* It is commonly used to replace or transform individual values.
* It is mainly applied to a single column.



**Demonstrate the Concept**

**Example:**

* The Senior column contains True and False values.
* map() can convert them into Yes and No.

**AI/ML Usage:**

* map() is used to transform categorical values into meaningful labels during data preprocessing.

```python
# Implement the Concept:
# map() - Senior col value change
employees["Senior"] = employees["Senior"].map({True: "Yes",False: "No"})
print(employees.head())
```

map() converts each True value into Yes and each False value into No

## **8. replace()**



**Understand the Concept**

* replace() is used to replace existing values in a Series or DataFrame.
* It can replace one value or multiple values.
* It is useful for correcting or standardizing data values.



**Demonstrate the Concept**

**Real-Life Example:**

* A company has inconsistent team names in its employee data.
* replace() can standardize those values.

**AI/ML Usage:**

* replace() is used to clean and standardize categorical values during data preprocessing.

```python
# Implement the Concept:
# replace() - Replace a team vales
employees["Team"] = employees["Team"].replace("Client Services","Customer Services")
print(employees.head())
```

**Interpretation**

* replace() changes the selected value in the Team column.
* Client Services is replaced with Customer Services.

## **9. Duplicate Handling**



**Understand the Concept**

* Duplicate rows are repeated records in a DataFrame.
* duplicated() is used to identify duplicate rows.
* drop_duplicates() is used to remove duplicate rows.



**Demonstrate the Concept**

**Real-Life Example:**

* A company may accidentally store the same employee record more than once.
* Duplicate handling helps remove repeated records before analysis.

**AI/ML Usage:**

* Duplicate handling helps prevent repeated records from affecting data analysis and Machine Learning models.

```python
# Implement the Concept:
# Find duplicate rows - return T or F then sum
employees.duplicated().sum()
```

Here no duplicate values in employees,so we add duplicates

```python
# Add duplicate row
employees.loc[0] = employees.loc[10]
employees.loc[1] = employees.loc[11]
print(employees.loc[0])
print(employees.loc[10])
print(employees.loc[1])
print(employees.loc[11])
```

```python
# Check duplicate
employees[employees.duplicated()]
```

```python
# drop_duplicates() - Remove duplicate value
employees = employees.drop_duplicates()
# Check duplicate
employees.duplicated().sum()
```

**Interpretation**

* duplicated() identifies repeated rows.
* Two duplicate rows were added.
* drop_duplicates() removes the repeated records.

## **Best Practices**

* Use clear and meaningful column names.
* Check the data before dropping rows or columns.
* Use assign() when you want to create a modified DataFrame.
* Use apply() and map() only when a transformation is required.
* Handle duplicate records before analysis.

## **Common Mistakes**

* Using the wrong column name while renaming or dropping.
* Accidentally removing required rows or columns.
* Confusing apply() and map().
* Modifying the original DataFrame unintentionally.
* Removing duplicates without checking whether they are valid repeated records.

## **Advantages**

* Makes data cleaning easier.
* Helps create and modify columns.
* Makes transformations simple.
* Helps remove unwanted and duplicate data.
* Useful for preparing data for analysis and Machine Learning.

## **Limitations**

* Incorrect transformations can change data incorrectly.
* apply() can be slower for large datasets.
* Dropping data permanently can cause information loss.
* Duplicate removal requires careful checking.
