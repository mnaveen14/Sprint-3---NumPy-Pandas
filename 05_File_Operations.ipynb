## **1. CSV Files**



**Understand the Concept**

* CSV means Comma-Separated Values.
* A CSV file stores data in rows and columns.
* It is a simple and commonly used file format for storing data.
* Pandas provides read_csv() to read a CSV file.
* We can use to_csv() to save a DataFrame as a CSV file.



**Demonstrate the Concept**

**Real-Life Example:**

* A weather application can save daily weather information in a CSV file.
* The saved file can later be opened and analyzed.

**AI/ML Usage:**

* CSV files are commonly used to store and load datasets for Data Science and Machine Learning.

```python
# Labraries:
import pandas as pd
```

```python
# Implement the Concept:
# weather data
weather = pd.DataFrame({"Date": ["2026-08-01", "2026-08-02", "2026-08-03"],
                        "Temperature": [30, 31, 29],
                        "Humidity": [72, 68, 75]})
```

```python
# Write data to CSV file
weather.to_csv("weather.csv", index=False)
```

```python
# Read the same CSV file
weather_data = pd.read_csv("weather.csv")
print(weather_data)
```

**Interpretation**

* to_csv() saves the DataFrame into a CSV file.
* read_csv() reads the same CSV file back into Pandas.

## **6. SQL Tables**



**Understand the Concept**

* SQL tables store data in rows and columns inside a database.
* Pandas can read data from a SQL table using read_sql().
* Pandas can also write a DataFrame to a SQL table using to_sql().
* This helps us move data between Pandas and a database.



**Demonstrate the Concept**

**Real-Life Example:**

* An online shopping company stores customer orders in a SQL database.
* Pandas can read the order data and analyze it.

**AI/ML Usage:**

* Pandas can load data from SQL databases for cleaning, analysis, and Machine Learning.

```python
# Implement the Concept:
import sqlite3

# order data
orders = pd.DataFrame({"Order_ID": [101, 102, 103],
                       "Product": ["Laptop", "Phone", "Tablet"],
                       "Amount": [55000, 25000, 18000]})

# database connection
connection = sqlite3.connect("orders.db")
```

```python
# Write DataFrame to SQL table
orders.to_sql("orders", connection, if_exists="replace", index=False)
```

```python
# Read the same SQL table
order_data = pd.read_sql("SELECT * FROM orders", connection)
print(order_data)

connection.close()
```

**Interpretation**

* to_sql() saves the DataFrame into a SQL table.
* read_sql() reads the SQL table into a DataFrame.


## **7. Reading Multiple Files**



**Understand the Concept**

* Pandas can read multiple files from a folder.
* We can use file paths and a loop to read each file.
* The data from different files can be combined into one DataFrame.
* This is useful when data is stored in separate files.



**Demonstrate the Concept**

**Real-Life Example:**

* A company stores monthly sales in separate CSV files.
* Pandas can read all the files and combine them for analysis.

**AI/ML Usage:**

* Used to combine data from multiple files before data cleaning and Machine Learning.


```python
# Implement the Concept:

# Monthly files
january = pd.DataFrame({"Product": ["Laptop", "Phone"],
                        "Sales": [55000, 25000]})

february = pd.DataFrame({"Product": ["Tablet", "Headphones"],
                         "Sales": [18000, 3000]})
```

```python
# Write multiple files in csv
january.to_csv("january.csv", index=False)
february.to_csv("february.csv", index=False)
```

```python
# Read multiple files
files = ["january.csv", "february.csv"] # path
data = [] # empty list

# Use for loop
for file in files:
    df = pd.read_csv(file)
    data.append(df) # file data store on empty list
```

```python
# Combine all files - using concatination method
combined_data = pd.concat(data, ignore_index=True)
print(combined_data)
```

**Interpretation**

* Each CSV file is read separately.
* pd.concat() combines the data from all files.
* The result is one DataFrame containing data from both files.

## **8. File Path Handling**

**Demonstrate the Concept**

**Real-Life Example:**

**AI/ML Usage:**

```python
# Implement the Concept:
from pathlib import Path

# file path
file_path = Path("C:/Users/iamna/Documents/Sprint 3 - NumPy & Pandas for AIML Engineers") / "weather.csv" 

# Check if the file exists
if file_path.exists():
    sales = pd.read_csv(file_path)
    print(sales)
else:
    print("File not found:", file_path)
```
