# **Notebook 12 – String Operations**

```python
# Import Labraries: 
import pandas as pd 
```

```python
# Load Dataset 
customers = pd.read_csv("string_operations_dataset.csv")
customers.info()
```

## **1. lower()**



**Understand the Concept**

* lower() converts all characters in a string to lowercase.
* In Pandas, .str.lower() is used to apply lower() to a column.



**Demonstrate the Concept**

**Real-Life Example:**

* Customer names or city names may contain uppercase and lowercase characters.
* Converting them to lowercase helps maintain a consistent format.

**AI/ML Usage:**

* lower() is useful during text preprocessing to standardize text before Natural Language Processing and Machine Learning.

```python
# Implement the Concept:
# lower() - convert Customer_Name col to lowercase 
customers["Cust_Name_Lower"] = customers["Customer_Name"].str.lower()
customers[["Customer_Name","Cust_Name_Lower"]].sample(5)
```

.str.lower() converts all characters in Customer_Name to lowercase.

## **2. upper()**



**Understand the Concept**

* upper() converts all characters in a string to uppercase.
* In Pandas, .str.upper() is used to apply upper() to a column.



**Demonstrate the Concept**

**Real-Life Example:**

* Customer names or city names may have different letter cases.
* Converting them to uppercase helps maintain a consistent format.

**AI/ML Usage:**

* upper() can be used during text preprocessing to standardize text before Natural Language Processing and Machine Learning.

```python
# Implement the Concept:
# upper() - convert Cust_Name_Lower to uppercase
customers["Customer_Name_Upper"] = (customers["Cust_Name_Lower"].str.upper())
print(customers[["Cust_Name_Lower", "Customer_Name_Upper"]].head())
```

.str.upper() converts all characters in Cust_Name_Lower to uppercase.

## **3. strip()**



**Understand the Concept**

* strip() removes unwanted spaces from the beginning and end of a string.
* In Pandas, .str.strip() is used to remove leading and trailing spaces from a column.



**Demonstrate the Concept**

**Real-Life Example:**

* Customer names or city names may contain extra spaces due to incorrect data entry.
* strip() removes these unwanted spaces and cleans the data.

**AI/ML Usage:**

* strip() helps clean text data before analysis and Machine Learning.

```python
# Implement the Concept:
# strip() - remove spaces from city col
customers["City_Clean"] = (customers["City"].str.strip())
print(customers[["City", "City_Clean"]].head())
```

 .str.strip() removes spaces from the beginning and end of City values.

## **4. split()**




**Understand the Concept**

* split() divides a string into multiple parts based on a separator.
* In Pandas, .str.split() is used to split string values in a column.


**Demonstrate the Concept**



**Real-Life Example:**

* A customer full name contains both first name and last name.
* split() can separate them into individual parts for analysis.

**AI/ML Usage:**

* split() helps extract useful information from text data during preprocessing.

```python
# Implement the Concept:
# split() -  Customer_Name into first name & last name
customers[["First_Name", "Last_Name"]] = (customers["Customer_Name"].str.strip().str.split(" ", expand=True))
print(customers[["Customer_Name", "First_Name", "Last_Name"]].head())
```

**Interpretation**

* .str.strip() removes extra spaces from Customer_Name.
* .str.split() separates the name based on the space.
* expand=True creates separate columns for each part.

## **5. contains()**



**Understand the Concept**

* contains() checks whether a string contains a specific word or pattern.
* In Pandas, .str.contains() is used to search for text in a column.
* It returns True or False for each value.



**Demonstrate the Concept**



**Real-Life Example:**

* A company can find all customer feedback containing the word delivery.
* This helps identify and analyze feedback related to delivery.

**AI/ML Usage:**

* contains() can be used to filter text data and identify specific keywords during text preprocessing.

```python
# Implement the Concept:
# contains() -  feedback containing word delivery
delivery_feedback = customers[customers["Feedback"].str.contains("delivery", case=False, na=False)]
print(delivery_feedback[["Customer_ID", "Customer_Name", "Feedback"]].head())
```

**Interpretation**

* .str.contains() checks whether Feedback contains the word delivery.
* case=False ignores uppercase and lowercase differences.
* na=False handles missing values by treating them as False.

## **6. replace()**



**Understand the Concept**

* replace() is used to replace specific text or values with new text or values.
* In Pandas, .str.replace() is used to replace text within string values.



**Demonstrate the Concept**



**Real-Life Example:**

* A company changes its email domain from company.com to example.com.

**AI/ML Usage:**

* replace() helps clean and standardize text data before data analysis and Machine Learning.

```python
# Implement the Concept:
# replace()  - company.com to example.com in Email col
customers["upd_email"] = (customers["Email"].str.replace("company.com","example.com",regex=False))
print(customers[["Email", "upd_email"]].head())
```

**Interpretation**

* .str.replace() searches for company.com in the Email column.
* Matching text is replaced with example.com.
* regex=False treats the search value as normal text.


## **7. Regular Expressions (Regex)**



**Understand the Concept**

* Regular Expressions, or Regex, are patterns used to search, match, extract, or replace text.
* In Pandas, Regex can be used with string methods such as .str.contains(), .str.extract(), and .str.replace().
* Regex is useful when text follows a specific pattern.



**Demonstrate the Concept**

**Real-Life Example:**

* A company wants to identify customers using gmail.com email addresses.
* Regex can search the Email column and filter records with gmail.com.

**AI/ML Usage:**

* Regex is useful for cleaning text, extracting patterns, and removing unwanted characters before Natural Language Processing and Machine Learning.

```python
# Implement the Concept:

# Regex -  find customers with gmail.com email  
gmail_customers = customers[customers["Email"].str.contains("@gmail\.com$",regex=True, na=False)]
print(gmail_customers[["Customer_ID", "Customer_Name", "Email"]].head(10))
```

**Interpretation**

* @gmail matches the gmail part of the email address.
* \. matches the dot character.
* com matches the domain extension.
* $ ensures that gmail.com appears at the end of the email address.

## **Best Practices**

* Use .str methods to perform string operations on Pandas columns.
* Handle missing values using parameters such as na=False when required.
* Use strip() to remove unwanted spaces.
* Standardize text using lower() or upper() when needed.
* Use Regex carefully for complex text patterns.
* Keep the original column when creating cleaned or transformed columns.


## **Common Mistakes**

* Applying string methods to non-string columns.
* Forgetting to handle missing values.
* Confusing replace() with .str.replace().
* Using an incorrect Regex pattern.
* Forgetting regex special characters such as . and $.
* Splitting data without checking the number of parts.

## **Advantages**

* Helps clean and standardize text data.
* Makes text filtering easier.
* Extracts useful information from string values.
* Supports pattern matching using Regex.
* Useful for data preprocessing and NLP.

## **Limitations**

* Complex Regex patterns can be difficult to understand.
* Incorrect patterns may produce unexpected results.
* String operations can be slower on very large datasets.
