# **Notebook 9 – Grouping & Aggregation**

## **1. groupby()**



**Understand the Concept**

* groupby() is used to divide data into groups based on one or more columns.
* It allows us to perform calculations separately for each group.
* It is commonly used with functions such as sum(), mean(), count(), and max().



**Demonstrate the Concept**

**Real-Life Example:**

* A food delivery company wants to compare order values across different cities.
* groupby() helps summarize the orders for each city.

**AI/ML Usage:**

* groupby() is used to create aggregated features and understand patterns within different groups.

```python
# Import Libraries: 
import pandas as pd 
```

```python
# Load Datasetabs
food_delivery = pd.read_csv("food_delivery_dataset.csv")
food_delivery.head()
```

```python
# Implement the Concept:
# groupby() - total order value for each city
city_sales = food_delivery.groupby("City")["Order_Value"].sum()
print(city_sales)
```

```python
# groupby() - max order value for each city 
food_delivery.groupby("City")["Order_Value"].max()
```

its shows the total order and max order value generated by each city

## **2. agg()**



**Understand the Concept**

* agg() is used to apply multiple aggregation functions to grouped data.
* It allows us to calculate different summaries such as sum(), mean(), and max() at the same time.
* It is useful when we need more than one calculation from grouped data.



**Demonstrate the Concept**

**Example:**

* Food delivery orders are grouped by city.
* agg() can calculate the total order value, average order value, and maximum order value for each city.

**AI/ML Usage:**

* agg() is used to create multiple aggregated features for data analysis and Machine Learning.

```python
# Implement the Concept:
# agg() - perform some statistics for each city 
city_stat = food_delivery.groupby("City")["Order_Value"].agg(["sum", "mean", "max"])
print(city_stat)
```

agg() applies multiple aggregation functions to the grouped data.

## **3. transform()**



**Understand the Concept**

* transform() performs a calculation within each group and returns a result with the same number of rows as the original data.
* It is useful when we want to add group-level information back to every row.
* Unlike groupby().agg(), transform() keeps the original row structure.



**Demonstrate the Concept**


**Real-Life Example:**

* A food delivery company wants to compare each order with the average order value of its city.
* transform() can add the city average to every order.

**AI/ML Usage:**

* transform() is used to create group-based features during feature engineering.

```python
# Implement the Concept:
# transform() - average order value for each city
food_delivery["City_Average_Order"] = (food_delivery.groupby("City")["Order_Value"]
                                                    .transform("mean"))
print(food_delivery.head())
```

**Interpretation**

* transform("mean") calculates the average Order_Value for each city.
* The city average is added to every order belonging to that city.

## **4. filter()**



**Understand the Concept**

* filter() is used with groupby() to keep only the groups that satisfy a condition.
* It checks each group based on a given rule.
* Groups that do not satisfy the condition are removed.



**Demonstrate the Concept**

**Example:**

* Food delivery orders are grouped by restaurant.
* filter() can keep only restaurants whose total order value meets a required condition.

**AI/ML Usage:**

* filter() can be used to select meaningful groups before creating aggregated features for Machine Learning.

```python
# Implement the Concept:
# filter() - show restaurants with more than 30 orders
filtered_data = food_delivery.groupby("Restaurant").filter(lambda group: len(group) > 30)

print(filtered_data.head())
```

**Interpretation**

* groupby("Restaurant") creates a group for each restaurant.
* filter() checks the number of orders in each restaurant.
* Only restaurants having more than 30 orders are kept.

## **5. Pivot Tables**



**Understand the Concept**

* A Pivot Table is used to summarize and analyze data based on one or more categories.
* It can calculate values such as sum, mean, count, and other aggregations.
* It helps compare grouped data in a table format.



**Demonstrate the Concept**

**Example:**

* Food delivery data contains cities and cuisines.
* A Pivot Table can show the total order value for each city and cuisine.

**AI/ML Usage:**

* Pivot Tables help explore relationships between categorical variables and numerical features before Machine Learning.

```python
# Implement the Concept:
# Create a Pivot Table
pivot_table = pd.pivot_table(food_delivery,
                             values="Order_Value",
                             index="City",
                             columns="Cuisine",
                             aggfunc="sum",
                             fill_value=0)

print(pivot_table)
```

it easy to compare order values across cities and cuisines

## **6. Crosstab**



**Understand the Concept**

* crosstab() is used to create a frequency table between two or more categorical variables.
* It shows how often combinations of category values occur.
* It is useful for comparing categorical data.



**Demonstrate the Concept**

**Example:**

* Food delivery data contains City and Payment information.
* crosstab() can show how many orders were made using each payment method in each city.

**AI/ML Usage:**

* crosstab() helps analyze relationships between categorical features before Machine Learning.

```python
# Implement the Concept:
# Create a crosstab
payment_by_city = pd.crosstab(food_delivery["City"],food_delivery["Payment"])
print(payment_by_city)
```

**Interpretation**

* crosstab() counts the combinations of City and Payment.
* Each row represents a city.
* Each column represents a payment method.
* The values show the number of orders for each combination.

## **Best Practices**

* Use groupby() to organize data into meaningful groups.
* Use agg() when multiple summary calculations are required.
* Use transform() when the grouped result needs to be returned to the original rows.
* Use filter() to keep only groups that meet a condition.
* Use Pivot Tables for easy comparison of grouped data.
* Use crosstab() to analyze relationships between categorical variables.

## **Common Mistakes**

* Grouping by the wrong column.
* Applying aggregation functions to inappropriate columns.
* Confusing agg() and transform().
* Using filter() with an incorrect condition.
* Confusing Pivot Tables with crosstab().
* Forgetting to handle missing values when required.

## **Advantages**

* Makes data summarization easier.
* Helps identify patterns between groups.
* Supports multiple aggregation functions.
* Useful for business and statistical analysis.
* Helps create features for Machine Learning.

## **Limitations**

* Complex grouping operations can be difficult to understand.
* Large datasets may require more memory and processing time.
* Incorrect grouping can produce misleading results.
* Aggregated results may hide individual-level information.
