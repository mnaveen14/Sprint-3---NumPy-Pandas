## **3. Series**



**Understand the Concept**

* A Series is a one-dimensional data structure in Pandas.
* It stores a list of values with an index.
* By default, the index starts from 0.
* A Series can store numbers, text, or other types of data also.



**Demonstrate the Concept**

**Real-Life Example:**

* A food delivery app can store the prices of different food items in a Series.
* Each price can be accessed using its index.

**AI/ML Usage:**

* Series are used to store and process individual columns of a dataset during data preparation.

```python
# Implement the Concept:
# Food item prices
prices = pd.Series([120, 180, 250, 90])

print("Food Prices:")
print(prices)
```

A Series is useful for working with a single column of data.

## **4. DataFrame**

**Understand the Concept**

* A DataFrame is a two-dimensional data structure in Pandas.
* It stores data in rows and columns.
* Each column can contain a different type of data.
* It is similar to a table with rows and columns.


**Demonstrate the Concept**

**Real-Life Example:**

* A shopping app can use a DataFrame to store product details such as name, price, category, and stock.

**AI/ML Usage:**

* DataFrames are used to store and prepare datasets before Machine Learning models are trained.

```python
# Implement the Concept:
# product data
data = {"Product": ["Laptop", "Phone", "Headphones"],
        "Price": [55000, 25000, 3000],
        "Stock": [10, 25, 50]}

products = pd.DataFrame(data)

print(products)
```

* Each row represents one product.
* Each column contains a specific type of information.

## **5. Creating DataFrames**



**Understand the Concept**

* A DataFrame can be created using the Pandas DataFrame function.
* We can create a DataFrame using a dictionary, list, or other data sources.
* Each key in a dictionary can become a column.
* Each row contains the related values.



**Demonstrate the Concept**

**Real-Life Example:**

* A grocery shop can create a DataFrame to store item names, prices, and available stock.


**AI/ML Usage:**

* DataFrames are used to organize raw data before cleaning and preparing it for Machine Learning.

```python
# Implement the Concept:
# Create DataFrame using a dictionary
data_dict = {"Product": ["Rice", "Milk", "Bread"],
             "Price": [60, 50, 40]}

df_dict = pd.DataFrame(data_dict)
print(df_dict)

```

```python
# Create DataFrame using a list
data_list = [["Rice", 60],
             ["Milk", 50],
             ["Bread", 40]]
df_list = pd.DataFrame(data_list, columns=["Product", "Price"])
print(df_list)
```

## **6. Data Types**



**Understand the Concept**

* Data type tells us what kind of data is stored in a column.
* Pandas supports data types such as integer, float, string, and boolean.
* We can use dtypes to check the data types of DataFrame columns.



**Demonstrate the Concept**

**Real-Life Example:**

* A food delivery app stores food names, prices, available quantities, and delivery status.
* Each type of information can have a different data type.

**AI/ML Usage:**

* Data types are checked and changed when preparing data for Machine Learning models.

```python
# Implement the Concept:
# product data
data = {"Product": ["Rice", "Milk", "Bread"],
        "Price": [60.5, 50.0, 40.5],
        "Quantity": [20, 15, 30],
        "Available": [True, True, False]}

products = pd.DataFrame(data)

# Check data types
print(products.dtypes)
```

dtypes shows the data type of each column.

## **7. Index**



**Understand the Concept**

* Index is used to identify each row in a DataFrame or Series.
* By default, Pandas gives numbers starting from 0 as the index.
* We can also create our own index values.
* Index helps us access and organize rows .



**Demonstrate the Concept**

**Real-Life Example:**

* A food delivery app can use an order ID as the index for each order.
* This makes it easier to identify a particular order.

**AI/ML Usage:**

* Index is used to identify and access specific rows while working with datasets.

```python
# Implement the Concept:
# order data
data = {"Food": ["Pizza", "Burger", "Biryani"],
        "Amount": [250, 150, 220]}
# Index
default_index_orders = pd.DataFrame(data)
own_index_orders = pd.DataFrame(data, index=["ORD101", "ORD102", "ORD103"])

print(default_index_orders)
print(own_index_orders)
```

## **8. Columns**



**Understand the Concept**

* Columns store a specific type of information in a DataFrame.
* Each column has a column name.
* We can select one or more columns from a DataFrame.
* Pandas provides columns to view the names of all columns.



**Demonstrate the Concept**

**Real-Life Example:**

* An online shopping app can have columns for product name, price, category, and stock.
* Each column stores one type of product information.

**AI/ML Usage:**

* Columns usually represent features that are used for data analysis and Machine Learning.

```python
# Implement the Concept:
# product data
data = {"Product": ["Laptop", "Phone", "Tablet"],
        "Price": [55000, 25000, 18000],
        "Stock": [10, 25, 15]}

products = pd.DataFrame(data)
```

## **9. Reading Data**



**Understand the Concept**

* Pandas can read data from different file formats.
* The read_csv() function is used to read CSV files.
* It loads the file data into a DataFrame.



**Demonstrate the Concept**

**Example:**

* A CSV file contains product prices and stock details.
* Pandas can read this data and create a DataFrame.

**AI/ML Usage:**

* Pandas is used to load datasets from files before cleaning and training Machine Learning models.

```python
# Implement the Concept:
# Read the CSV file
products = pd.read_csv("products.csv")

print(products.head(10)) 
```

```python
# Read Excel file
products = pd.read_excel("products.xlsx")

print(products.tail(10))
```

```python
# Read JSON file
products = pd.read_json("products.json")

print(products.sample(10))
```

## **10. Writing Data**



**Understand the Concept**

* Pandas can save DataFrame data into different file formats.
* We can save data as CSV, Excel, JSON, and Pickle files.
* Functions like to_csv(), to_excel(), to_json(), and to_pickle() are used to save data.
* Writing data is useful when we want to store our processed data for later use.



**Demonstrate the Concept**

**Real-Life Example:**

* An online shop can save its updated product data after making changes.
* The saved file can be used later for reporting or analysis.

**AI/ML Usage:**

* Processed datasets can be saved and reused for future analysis and Machine Learning tasks.

```python
# Implement the Concept:
# product data
data = {"Product": ["Rice", "Milk", "Bread"],
        "Price": [60, 50, 40]}

products = pd.DataFrame(data)

# Save data in different formats
products.to_csv("products_output.csv", index=False)
products.to_excel("products_output.xlsx", index=False)
products.to_json("products_output.json", orient="records")
```

```python
# View these files
# CSV file
csv_data = pd.read_csv("products_output.csv")
print("CSV File:")
print(csv_data)

# Excel file
excel_data = pd.read_excel("products_output.xlsx")
print("\nExcel File:")
print(excel_data)

# JSON file
json_data = pd.read_json("products_output.json")
print("\nJSON File:")
print(json_data)
```

## **Best Practices**

* Use clear column names when creating a DataFrame.
* Check the data types before working with the data.
* Use the correct function for each file format.
* Keep a backup of important data before making changes.
* Use index=False when saving a DataFrame if the index is not needed.

## **Common Mistakes**

* Using the wrong file path.
* Using the wrong function for a file type.
* Forgetting to install or import the required library.
* Saving the DataFrame with an unwanted index column.
* Using incorrect column names while accessing data.
