## **4. Creating Arrays**



**Understand the Concept**

- NumPy provides different ways to create arrays.
- We can create an array using `np.array()`.
- We can also create arrays filled with zeros or ones.
- We can create a range of numbers using `np.arange()`.





```python
# Implement the Concept:
# Creating a array in different ways
# using values
marks = np.array([70, 80, 90])

# array of zeros
zeros = np.zeros(3)

# array of ones
ones = np.ones(3)

#  array of range
numbers = np.arange(1, 6)

print("Marks:", marks)
print("Zeros:", zeros)
print("Ones:", ones)
print("Numbers:", numbers)
```

## **5. Dimensions**



**Understand the Concept**

- Dimension tells us how many levels an array has.
- 1D array -  single row of values.
- 2D array - rows and columns.
- 3D array - contain multiple 2D arrays.




**Demonstrate the Concept**

**Example:**

* 1D array: [10, 20, 30, 40]
* 2D array: [[10, 20], [30, 40]]
* 3D array: [[[1,2],[3]],[[4],[5,6]]


**AI/ML Example:**

ML data is often stored as a 2D array where rows represent data records and columns represent features.



```python
# Implement the Concept:
# Create a 1D array
one_d = np.array([10, 20, 30, 40])

# Create a 2D array
two_d = np.array([[10, 20], [30, 40]])

# Create a 3D array
three_d = np.array([[[10,20],[30,40]],
                    [[40,50],[60,70]]])

print("1D Array:", one_d)
print("\n2D Array:\n",two_d)
print("\n3D Array:\n",three_d)
```

## **8. Data Types**



**Understand the Concept**

## **9. Indexing**



**Understand the Concept**

- Indexing means accessing a specific value from an array.
- Indexing number starts from 0.



**Demonstrate the Concept**

**Real-Life Example:**

- Suppose a shopping list contains: ["Rice", "Milk", "Eggs", "Bread"]
- If we want the third item, we use index 2 because indexing starts from 0.

**AI/ML Usage:**

## **10. Slicing**



**Understand the Concept**

- Slicing means getting a part of an array.
- We can select multiple values using a start and end index.
- The basic format is `array[start:end:step]`.




**Demonstrate the Concept**

**Real-Life Example:**

- Student's weekly study hours 

**AI/ML Usage:**

- Used to select a part of an array when preparing or processing data.



```python
# Implement the Concept:
# weekly study hours data
study_hours = np.array([2, 3, 4, 5, 3, 6, 4])

print("last 3 days :", study_hours[4:])
print("day 4 & 5 :", study_hours[3:5])
print("even days :", study_hours[1::2])
```

## **11. Reshaping**



**Understand the Concept**

- Reshaping means changing the structure of an array.
- It changes the number of rows and columns without changing the values.



**Demonstrate the Concept**

**Real-Life Example:**

* A teacher has marks of 6 students stored in one row.
* The marks can be reshaped into 2 rows and 3 columns to organize the data better.

```python
# Implement the Concept:
# Marks data - in 1D 
marks = np.array([10, 20, 30, 40, 50, 60])

# Reshape into 2D array
reshaped_marks = marks.reshape(2, 3)

print("Original Array:", marks)
print("Original Shape:", marks.shape)

print("\nReshaped Array:\n",reshaped_marks)
print("New Shape:", reshaped_marks.shape)
```

## **12. Flatten**



**Understand the Concept**

- Flatten means converting a multi-dimensional array into a one-dimensional array..
- It keeps all the values but removes the rows and columns.




**Demonstrate the Concept**

**Real-Life Example:**

A school stores marks of students in rows and columns. If we want all the marks in a single list, we can use flatten.

**AI/ML Usage:**

- Used to convert multi-dimensional data into a single array before giving it to a Machine Learning model.

```python
# Implement the Concept:
# marks data - 2d into 1d
marks = np.array([[80, 75, 90],
                  [85, 88, 92]])

# Convert the 2D array into a 1D array
flat_marks = marks.flatten()

print("Original Array:\n", marks)
print("\nFlattened Array:\n",flat_marks)
```

## **13. Copy vs View**

**Understand the Concept**

**Copy:**
- A copy creates a separate array with the same values.
- Changes made to a copy do not affect the original array.

**View:**

- A view uses the same data as the original array.
- Changes made to a view can also change the original array.



**Demonstrate the Concept**

**Real-Life Example:**

**Copy:**

If you photocopy a document, changes to the photocopy do not affect the original document. This is similar to a copy.

**View:**

If two people use the same original document, a change made by one person can affect what the other person sees. This is similar to a view.

**AI/ML Usage:**

- Used to control how data is shared and changed during data preprocessing.


```python
# Implement the Concept:
# Data
original = np.array([10, 20, 30])

# copy
copied_array = original.copy()
copied_array[0] = 100

print("Original after Copy Change:", original)
print("Copied Array:", copied_array)
```

```python
# view
view_array = original.view()
view_array[1] = 200

print("Original after View Change:", original)
print("View Array:", view_array)
```

## **14. Broadcasting**



**Understand the Concept**

- Broadcasting allows arrays with different shapes to perform arithmetic operations automatically.



**Demonstrate the Concept**

**Real-Life Example:**

- A shop has three product prices.
- If the shop increases all prices by ₹10, we can add `10` to the complete array at once.

**AI/ML Usage:**

- Used to apply the same value or calculation to many data values during data preprocessing.



```python
# Implement the Concept:
# product prices
prices = np.array([100, 200, 300])

# Add 10 using broadcasting
new_prices = prices + 10

print("Original Prices:", prices)
print("New Prices:", new_prices)
```

## **15. Array Operations**



**Understand the Concept**

- NumPy allows us to perform calculations directly on arrays.
- We can add, subtract, multiply, and divide array values.





**Demonstrate the Concept**

**Real-Life Example:**

A shop has the quantity and price of three products:

- Quantity: `[2, 3, 4]`
- Price: `[100, 200, 150]`

We can multiply both arrays to calculate the total price for each product.

**AI/ML Usage:**

- Used for fast calculations on numerical data during data processing and feature preparation.


```python
# Implement the Concept:
# data
quantity = np.array([2, 3, 4])
price = np.array([100, 200, 150])

# array operations
total_price = quantity * price
price_after_discount = price - 20

print("Quantity:", quantity)
print("Price:", price)
print("Total Price:", total_price)
print("Price After Discount:", price_after_discount)
```

## **16. Universal Functions (ufuncs)**



**Understand the Concept**

- ufuncs are  NumPy functions used to perform calculations on array values.
- A ufunc can work on every value in an array at the same time.
- Common ufuncs include np.sqrt(), np.abs(), np.exp() and np.sin().






**Demonstrate the Concept**

**Real-Life Example:**

- A shop records the difference between expected sales and actual sales.
- Using np.abs(), we can find the difference without considering whether it is positive or negative.

**AI/ML Usage:**

- Used to apply mathematical functions to large amounts of numerical data during data processing and model preparation.

```python
# Implement the Concept:
# data
arr = np.array([1,2,3])

# Universal Functions
print(np.square(arr))
print(np.exp(arr))
print(np.log(arr))
print(np.sin(arr))
print(np.cos(arr))
print(np.abs([-2,-5,6]))
```

## **Common Mistakes**

- Forgetting that array indexing starts from `0`.
- Using an incorrect shape while reshaping an array.
- Trying to reshape an array into a size that does not match the total number of values.
- Confusing `copy()` and `view()`.
- Using arrays with incompatible shapes during broadcasting.

## **Advantages**

- Fast numerical calculations.
- Easy to work with large numerical data.
- Supports one-dimensional and multi-dimensional arrays.
- Provides many mathematical functions.
- Useful for Data Science, AI, and Machine Learning.

## **Limitations**

- Mainly designed for numerical data.
- Requires understanding of arrays and dimensions.
- Not as convenient as Pandas for working with labelled table data.
- Some operations require understanding of array shapes.
