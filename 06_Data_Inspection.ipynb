# **Notebook 6 – Data Inspection**

## **1. head()**



**Understand the Concept**

* head() is used to view the first few rows of a DataFrame.
* By default, it shows the first 5 rows.
* We can also give a number to view more rows.



**Demonstrate the Concept**

**Example:**

* An employee dataset contains many employee records.
* head() shows the first few employee records without displaying the complete dataset.


**AI/ML Usage:**

* head() is used to quickly inspect a dataset before data cleaning and model building.

```python
# Libraries 
import pandas as pd
# Read - employee data
employees = pd.read_csv("employees_dataset.csv")
```

```python
# head() - 1st 5 rows is default
print(employees.head())
```

```python
# head() - own value
print(employees.head(8))
```

**Interpretation**

* It helps us quickly check the beginning of the dataset.

## 2. tail()



**Understand the Concept**

* tail() is used to view the last few rows of a DataFrame.
* By default, it shows the last 5 rows.
* We can give a number inside tail() to view a specific number of rows.
* It is useful for checking the end of a dataset.



**Demonstrate the Concept**

**Real-Life Example:**

* A company receives new employee records regularly.
* tail() helps check the latest records at the end of the dataset.

**AI/ML Usage:**

* tail() is used to check the end of a dataset and make sure the data was loaded correctly.

```python
# tail() - last 5 rows
print(employees.tail())
```

```python
# tail() - last 7 rows
print(employees.tail(7))
```

It helps us quickly check the end of the dataset.

## **3. sample()**



**Understand the Concept**

* sample() is used to select random rows from a DataFrame.
* By default, it selects one random row.
* We can give a number inside sample() to select multiple random rows



**Demonstrate the Concept**

**Real-Life Example:**

* A company has a large employee dataset.
* sample() helps check random employee records without viewing the complete dataset.

**AI/ML Usage:**

* sample() is used to randomly inspect or select data during data analysis and Machine Learning.

```python
# Implement the Concept:
# sample() - random 5 rows
print(employees.sample(5))
```

**Interpretation**

* sample(5) selects 5 random rows from the employee dataset.
* The selected rows may be different each time the code runs.

## **4. info()**



**Understand the Concept**

* info() gives a quick summary of a DataFrame.
* It shows the number of rows and columns.
* It shows the column names and their data types.
* It also shows how many non-null values are present in each column.
* It is useful for checking the overall structure of a dataset.



**Demonstrate the Concept**

**Real-Life Example:**

* A company receives an employee dataset from another system.
* info() helps check the dataset structure before starting analysis.

**AI/ML Usage:**

* info() is used to check data types and missing values before data preprocessing.

```python
# Implement the Concept:
# info() - dataset information
employees.info()
```

**Interpretation**

* info() shows that the dataset has 500 rows and 6 col.
* It shows the data type of each column.
* It also shows the number of non-null values in each column.
* Gender and Team contain some missing values.

## **10. nunique()**



**Understand the Concept**

* nunique() is used to count the no of unique values in a column.
* It tells us how many different values are present.
* By default, missing values are not counted.
* It can be used for one column or for all columns.



**Demonstrate the Concept**

**Real-Life Example:**

* A company has employees working in different teams.
* nunique() helps find how many different teams are in the dataset.

**AI/ML Usage:**

* nunique() helps understand categorical features before data preprocessing and Machine Learning.

```python
# Implement the Concept:
# nunique() - 1 col
print("Unique Teams:", employees["Team"].nunique())
```

```python
# nunique() - all col
print(employees.nunique())
```

 nunique() counts the different values in a column.

## **11. value_counts()**



**Understand the Concept**

* value_counts() counts how many times each unique value appears in a column.
* It is mainly used for categorical data.
* It shows each value along with its count.
* By default, the results are arranged from highest count to lowest count.


**Demonstrate the Concept**

**Real-Life Example:**

* A company has employees working in different teams.
* value_counts() helps understand how employees are distributed across teams.

**AI/ML Usage:**

* value_counts() is used to understand the distribution of categorical features before Machine Learning.

```python
# Implement the Concept:
# value_counts() - Count employees in each team
team_counts = employees["Team"].value_counts()
print(team_counts)
```

value_counts() shows the no of employees in each team.

## **12. Dataset Inspection Practice**

* After learning the main inspection functions, we can use them together to understand a dataset.
* These functions help us check the rows, columns, data types, memory, and value distribution.
* This is an important first step before cleaning or analyzing data.

**AI/ML Usage:**

* Dataset inspection helps identify the structure and quality of data before Machine Learning.

**Notes**

* head() shows the beginning of the dataset.
* tail() shows the end of the dataset.
* sample() shows random rows.
* shape shows the number of rows and columns.
* columns shows the column names.
* dtypes shows the data types.
* nunique() shows the number of unique values.
* value_counts() shows how frequently each value appears.

## **Best Practices**

* Use head() and tail() to quickly check the dataset.
* Use info() to check data types and missing values.
* Use describe() to understand numerical data.
* Use shape and columns to understand the dataset structure.
* Use value_counts() to understand categorical data.

## **Common Mistakes**

* Confusing shape with columns.
* Using value_counts() on the wrong column.
* Forgetting that sample() selects random rows.
* Assuming describe() summarizes text columns by default.
* Ignoring missing values shown by info().

## **Advantages**

* Helps understand a dataset quickly.
* Makes it easier to find missing values and data types.
* Helps identify unique and repeated values.
* Useful before data cleaning and analysis.

## **Limitations**

* Inspection functions do not clean or modify the data.
* describe() mainly focuses on numerical columns by default.
* Large datasets can produce large outputs if too much data is displayed.
