## **1. Random Numbers**



**Understand the Concept**

* Random numbers are values generated by a computer in a random way.
* NumPy provides np.random.rand() to generate random numbers between 0 and 1.
* Every time we run the code, the values can be different.



**Demonstrate the Concept**

**Real-Life Example:**

* A game can generate random numbers to decide a player's chance or position.
* This makes the result different each time.

**AI/ML Usage:**

* Random numbers are used to create random data and initialize values in Machine Learning.
* They are also used when splitting data or testing models.

```python
# Labraries;
import numpy as np
```

```python
# Implement the Concept:
# Random num
random_numbers = np.random.rand(5)

print("Random Numbers:", random_numbers)
```

Five random numbers are generated between 0 and 1.

## **2. Random Seed**



**Understand the Concept**

* A random seed is used to control random number generation.
* NumPy provides np.random.seed() to set a starting point for random values.
* Using the same seed gives the same random values every time.



**Demonstrate the Concept**

**Real-Life Example:**

* A teacher randomly selects students for a group.
* Using the same starting rule can produce the same selection again.

**AI/ML Usage:**

* Random seeds are used to get the same results when testing and training Machine Learning models.

```python
# Implement the Concept:
# Set the random seed
np.random.seed(10)

# random numbers
random_numbers = np.random.rand(3)

print("Random Numbers:", random_numbers)
```

Running the same code again gives the same values.

## **3. Random Integers**



**Understand the Concept**

* Random integers are whole numbers generated randomly.
* NumPy provides np.random.randint() to generate random integers.
* We can set a range and the number of values we want.



**Demonstrate the Concept**

**Example:**

* Generate random integers from 1 to 10.
* The result can contain values such as 3, 7, 2, and 9.

**AI/ML Usage:**

* Random integers are used to create sample data and randomly select values during Machine Learning tasks.

```python
# Implement the Concept:
# Random Integers
random_int = np.random.randint(1,11) # start,stop(n-1),size
print("Random Integers:",random_int)
```

## **4. Random Choice**



**Understand the Concept**

* Random choice selects values randomly from an array.
* NumPy provides np.random.choice() for random selection.
* We can select one or more values from the given data.



**Demonstrate the Concept**

**Real-Life Example:**

* A teacher randomly selects a student for a classroom activity.
* The student is selected from the available students.

**AI/ML Usage:**

* Random choice is used to select random samples from a dataset.
* It can also be used in testing and data sampling.

```python
# Implement the Concept:
# students data
students = np.array(["Arun", "Bala", "Kavi", "Ravi", "Siva"])

# Select one student randomly
selected_student = np.random.choice(students)

print("Selected Student:", selected_student)
```

Selected student can be different each time.

## **5. Shuffle**



**Understand the Concept**

* Shuffle changes the order of values in an array randomly.
* NumPy provides np.random.shuffle() to shuffle an array.
* It changes the original array.




**Demonstrate the Concept**

**Real-Life Example:**

* A teacher shuffles the list of students before selecting teams.
* This gives everyone a random position.

**AI/ML Usage:**

* Shuffling is used to change the order of training data before training a Machine Learning model.
* It helps reduce the effect of the original data order.

```python
# Implement the Concept:
# Student data
students = np.array(["Arun", "Bala", "Kavi", "Ravi", "Siva"])

# Shuffle 
np.random.shuffle(students)
print("Shuffled Students:", students)
```

**Interpretation**

* The order of the students is changed randomly.
* np.random.shuffle() changes the original array.
* The order can be different each time the code runs.

## **6. Permutation**



**Understand the Concept**

* Permutation creates a random order of values.
* NumPy provides np.random.permutation() for this.
* Unlike shuffle(), permutation does not change the original array.
* It returns a new array with the values in a random order.



**Demonstrate the Concept**
**Real-Life Example:**

* A teacher has a list of students and wants to create a random order for a presentation.
* The original student list can be kept unchanged.

**AI/ML Usage:**

* Permutation is used to randomly reorder data during data preparation and testing.
* It can also be used to check the importance of features in Machine Learning.

```python
# Implement the Concept:
# students data
students = np.array(["Arun", "Bala", "Kavi", "Ravi", "Siva"])

# random permutation
random_order = np.random.permutation(students)  # not change og data

print("Original Students:", students)
print("Random Order:", random_order)
```

**Interpretation**

* The original array remains unchanged.
* np.random.permutation() creates a new array with a random order.
* The order can be different each time the code runs.

## **7. Random Distributions**



**Understand the Concept**

* A random distribution shows how random values are spread.
* NumPy provides different distributions for different types of data.

**Common Random Distributions**

**1. Normal Distribution**

* Most values are close to the average.
* Example: Student marks in a class.

**2. Uniform Distribution**

* All values in a given range have an equal chance.
* Example: Rolling a dice.

**3. Binomial Distribution**

* Shows the number of successes in a fixed number of trials.
* Example: Number of heads when tossing a coin several times.



**Demonstrate the Concept**

**AI/ML Usage:**

* Random distributions are used to generate sample data and initialize values in Machine Learning models.
* They are also useful for simulations and testing.

```python
# Implement the Concept:
# normal distribution
normal_data = np.random.normal(loc=50, scale=10, size=5) # loc - mean, scale - std

# uniform distribution
uniform_data = np.random.uniform(low=10, high=20, size=5)

# binomial distribution
binomial_data = np.random.binomial(n=10, p=0.5, size=5) # n - trials , p - probability

print("Normal Distribution:", normal_data)
print("Uniform Distribution:", uniform_data)
print("Binomial Distribution:", binomial_data)
```

## **Best Practices**

* Use a seed when you need the same random results.
* Use the correct random function for the required task.
* Use shuffle when you want to change the original array.
* Use permutation when you want a new randomly ordered array.
* Choose the distribution based on the type of data you need.

## **Common Mistakes**

* Forgetting to set a seed when repeatable results are required.
* Confusing shuffle and permutation.
* Using an incorrect range with randint().
* Expecting the same random values without setting a seed.
* Using the wrong distribution for the problem.

## **Advantages**

* Easy way to generate random data.
* Useful for testing and simulations.
* Helps in data sampling.
* Supports different random distributions.
* Useful in Machine Learning and Data Science.

## **Limitations**

* Random results can change each time the code runs.
* Random data may not always represent real-world data.
* Choosing the wrong distribution can give misleading results.
* Random generation can require more memory for large datasets.
