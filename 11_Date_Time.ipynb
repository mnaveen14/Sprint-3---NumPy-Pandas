# **Notebook 11 – Date & Time Operations**

```python
# Import Libraries 
import pandas as pd 
```

```python
# Load dataset 
orders = pd.read_csv("order_datetime_dataset.csv")
orders.info()
```

## **1. Datetime**



**Understand the Concept**

* Datetime represents both a date and a time.
* In Pandas, datetime data is commonly handled using pd.to_datetime



**Demonstrate the Concept**

**Real-Life Example:**

* An e-commerce company stores the date and time when each order is placed.
* Datetime operations help analyze orders based on date and time.

**AI/ML Usage:**

* Datetime data can be converted into useful features such as year, month, day, and hour for Machine Learning models.

```python
# Implement the Concept:
# Convert Order_Date & Delivery_Date datatype(datetime)
# Check current datatype
print(orders["Order_Date"].dtype)
print(orders["Delivery_Date"].dtype)
```

```python
# convert datatype
orders["Order_Date"] = pd.to_datetime(orders["Order_Date"])
orders["Delivery_Date"] = pd.to_datetime(orders["Delivery_Date"])

# Check changed datatype
print(orders["Order_Date"].dtype)
print(orders["Delivery_Date"].dtype)
```

```python
# Check current data & time 
from datetime import datetime
print(datetime.now())
```

**Interpretation**

* Initially, Order_Date & Delivery_Date is stored as an object data type.
* pd.to_datetime() converts into datetime64 format.

## **2. Timestamp**



**Understand the Concept**

* A Timestamp represents a specific date and time.
* In Pandas, pd.Timestamp() is used to create a Timestamp object.
* Timestamp is useful for working with individual date and time values.



**Demonstrate the Concept**

**Real-Life Example:**

* An order is placed at a specific date and time.
* A Timestamp can represent the exact time when the order was created.

**AI/ML Usage:**

* Timestamps can be used to calculate time differences and create time-based features for Machine Learning.

```python
# Implement the Concept:
# Create Timestamp
order_time = pd.Timestamp("2026-08-20")
print(order_time)

# Check the data type
print(type(order_time))
```

**Interpretation**

* pd.Timestamp() creates a single date value.
* The Timestamp stores both the date and time together.

## **3. Date Parsing**



**Understand the Concept**

* Date parsing is the process of converting date values stored as strings into datetime format.
* pd.to_datetime() is commonly used for date parsing.
* The format parameter can be used when the date follows a specific format.



**Demonstrate the Concept**


**Real-Life Example:**

* Different systems may store dates in formats such as DD-MM-YYYY or YYYY/MM/DD.
* Date parsing converts these values into a consistent datetime format for analysis.

**AI/ML Usage:**

* Date parsing helps prepare date columns before extracting useful time-based features for Machine Learning.

```python
# Implement the Concept:
# date values in string format
dates = pd.Series(["20-08-2026", "21-08-2026", "22-08-2026"])

# Date Parsing -  string into datetime 
parsed_dates = pd.to_datetime( dates,format="%d-%m-%Y")
print(parsed_dates)
```

**Interpretation**

* The original date values are stored as strings.
* pd.to_datetime() converts them into datetime64 format.

## **4. Timedelta**



**Understand the Concept**

* Timedelta represents the difference between two dates or times.
* In Pandas, pd.Timedelta() can create a time duration.



**Demonstrate the Concept**

**Real-Life Example:**

* A delivery is delayed by 2 days due to bad weather or a logistics issue.
* pd.Timedelta(days=2) can be added to the original Delivery_Date to calculate the delayed delivery date.

**AI/ML Usage:**

* Time differences can be converted into useful features such as delivery duration for Machine Learning models.

```python
# Implement the Concept:
# Delivery delay - so add 2 days from accual delivey days
orders["Delayed_Delivery"] = orders["Delivery_Date"] + pd.Timedelta(days = 2)
orders.head()
```

**Interpretation**

* pd.Timedelta(days=2) represents a duration of 2 days.
* The 2-day duration is added to Delivery_Date.

## **5. Time Zones**



**Understand the Concept**

* A time zone represents the local time of a specific geographical region.
* The same date and time can have different values in different time zones.
* Pandas can assign and convert time zones using tz_localize() and tz_convert().



**Demonstrate the Concept**

**Example:**

* An order time is recorded in the Asia/Kolkata time zone.
* The same time can be converted to another time zone such as America/New_York.

**Real-Life Example:**

* An international e-commerce company receives orders from customers in different countries.
* Time zones help convert order times into the local time of each region.

**AI/ML Usage:**

* Time zone conversion helps standardize time-based data collected from different geographical locations before analysis and Machine Learning.

```python
# Implement the Concept:
# Assign India time zone
orders["Order_Date_IST"] = orders["Order_Date"].dt.tz_localize("Asia/Kolkata")

# Convert IST time to US Eastern time
orders["Order_Date_US"] = orders["Order_Date_IST"].dt.tz_convert("America/New_York")

print(orders[["Order_Date_IST", "Order_Date_US"]].head())
```

**Interpretation**

* tz_localize() assigns the Asia/Kolkata time zone to Order_Date.
* tz_convert() converts the time from Asia/Kolkata to America/New_York.

## **6. Extracting Date Features**



**Understand the Concept**

* Date features can be extracted from a datetime column.
* Common features include year, month, day, weekday, and hour.
* Pandas uses the .dt accessor to extract these features.



**Demonstrate the Concept**

**Real-Life Example:**

* An e-commerce company wants to identify which month, day, or hour receives the highest number of orders.


**AI/ML Usage:**

* Extracted features such as month, weekday, and hour can be used as input features for Machine Learning models.

```python
# Implement the Concept:
# Extract date and time features
orders["Year"] = orders["Order_Date"].dt.year
orders["Month"] = orders["Order_Date"].dt.month
orders["Day"] = orders["Order_Date"].dt.day
orders["Weekday"] = orders["Order_Date"].dt.day_name()
orders["Hour"] = orders["Order_Date"].dt.hour
print(orders[["Order_Date", "Year", "Month", "Day", "Weekday", "Hour"]].head())
```

```python
# highest orders by month, day 
print("Month :", orders["Month"].value_counts().idxmax())
print("Day :", orders["Day"].value_counts().idxmax())
```

**Interpretation**

* value_counts() - counts the no of orders for each month & day.
* idxmax() - returns the month & day with the highest no of orders.


## **Best Practices**

* Convert date columns to datetime format before performing date operations.
* Use pd.to_datetime() for date parsing.
* Use pd.Timestamp() for individual date and time values.
* Use pd.Timedelta() for adding or subtracting time durations.
* Handle missing date values before performing calculations.
* Use appropriate time zones when working with international data.
* Extract useful date features for analysis and Machine Learning.

## **Common Mistakes**

* Performing date operations on string columns.
* Using an incorrect date format during parsing.
* Confusing Timestamp with Timedelta.
* Using tz_convert() before assigning a time zone.
* Forgetting to handle missing values in date columns.
* Extracting unnecessary date features for Machine Learning.

## **Advantages**

* Makes time-based analysis easier.
* Helps calculate durations and delays.
* Supports data from different time zones.
* Allows extraction of useful features.
* Helps identify patterns based on time.
