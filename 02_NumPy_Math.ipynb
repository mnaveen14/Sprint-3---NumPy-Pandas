# **Notebook 2 – NumPy Mathematical Operations**

## **1. Sum**



**Understand the Concept**

* Sum means adding all the values together.
* NumPy provides np.sum() to quickly add values in an array.



**Demonstrate the Concept**

**Real-Life Example:**

* A shop sells 10, 20, 30, and 40 products on four days.
* The sum gives the total number of products sold.

**AI/ML Usage:**

* Sum is used to calculate totals in datasets.
* It can also be used in statistical calculations and feature processing.

```python
# Libraries
import numpy as np
```

```python
# Implement the Concept:
# sales data
sales = np.array([10, 20, 30, 40])

# Sum - total sales
total_sales = np.sum(sales)

print("Total Products Sold:", total_sales)
```

## **2. Mean**

## **4. Variance**



**Understand the Concept**

* Variance shows how much the values are spread out from the mean.
* NumPy provides np.var() to calculate variance.
* small variance -  values are close to the mean.
* large variance -  values are more spread out.

  
**Formula :**

 Variance = Average of the squared differences from the mean.



**Demonstrate the Concept**

**Real-Life Example:**

* A teacher compares the marks of students in a class.
* Variance shows how much the marks differ from the average mark.

**AI/ML Usage:**

* Variance is used to understand how features are spread in a dataset.

```python
# Implement the Concept:
# Student marks
marks = np.array([10, 20, 30])

# variance
variance = np.var(marks)

print("Variance:", variance)
```

## **5. Standard Deviation**



**Understand the Concept**

* Standard deviation shows how far the values are from the mean.
* NumPy provides np.std() to calculate standard deviation.
* small standard deviation - the values are close to the mean.
* large standard deviation - the values are more spread out.

  
**Formula**

* Standard Deviation = Square root of Variance



**Demonstrate the Concept**

**Real-Life Example:**

* A teacher checks the marks of students in a class.
* Standard deviation shows how much the marks differ from the average mark.

**AI/ML Usage:**

* Standard deviation is used to understand the spread of features in a dataset.
* It is also used in data scaling and preprocessing.

```python
# Implement the Concept:
# Student marks
marks = np.array([10, 20, 30])

# standard deviation
standard_deviation = np.std(marks)

print("Standard Deviation:", standard_deviation)
```

## **11. Dot Product**



**Understand the Concept**

* Dot product means multiplying the matching values of two arrays and adding the results.
* NumPy provides np.dot() to calculate the dot product.
* Both arrays should have compatible sizes.

**Formula**

* Dot Product = (a₁ × b₁) + (a₂ × b₂) + ... + (aₙ × bₙ)



**Demonstrate the Concept**

**Example:**

* First array: 2, 3, 4
* Second array: 5, 6, 7
* Dot product = (2 × 5) + (3 × 6) + (4 × 7) = 56

**AI/ML Usage:**

* Dot product is widely used in Machine Learning to calculate weighted sums and is a basic operation in neural networks.

```python
# Implement the Concept:
# two arrays
array_one = np.array([2, 3, 4])
array_two = np.array([5, 6, 7])

# dot product
result = np.dot(array_one, array_two)

print("Dot Product:", result)
```

## **12. Matrix Multiplication**



**Understand the Concept**

* Matrix multiplication means multiplying two matrices to produce a new matrix.
* NumPy provides np.matmul() for matrix multiplication.
* The no of columns in the first matrix must match the no of rows in the second matrix.





**Demonstrate the Concept**

**Example:**

* First matrix: [[1, 2], [3, 4]]
* Second matrix: [[5, 6], [7, 8]]
* The result is [[19, 22], [43, 50]].



**AI/ML Usage:**

* Matrix multiplication is used in neural networks to calculate outputs from inputs and weights.

```python
# Implement the Concept:
# two matrices
matrix_a = np.array([[1, 2],
                     [3, 4]])

matrix_b = np.array([[5, 6],
                     [7, 8]])

# matrix multiplication
result = np.matmul(matrix_a, matrix_b)

print("Matrix Multiplication:")
print(result)
```

## **13. Transpose**



**Understand the Concept**

* Transpose changes the rows of a matrix into columns and the columns into rows.
* NumPy provides np.transpose() to transpose an array.
* The values stay the same, but their positions change.



**Demonstrate the Concept**

**Example:**

* Matrix: [[1, 2, 3], [4, 5, 6]]
* Transpose: [[1, 4], [2, 5], [3, 6]]



**AI/ML Usage:**

* Transpose is used to change the shape and arrangement of matrices during Machine Learning calculations.

```python
# Implement the Concept:
# matrix
matrix = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

# Transpose - convert row to col & col to row
transposed_matrix = np.transpose(matrix)

print("Original Matrix:\n", matrix)
print("Shape :", matrix.shape)
print("\nTransposed Matrix:\n",transposed_matrix)
print("Shape :", transposed_matrix.shape)
```

## **14. Basic Linear Algebra Operations**



**Understand the Concept**

* Linear algebra deals with vectors, matrices, and mathematical operations.
* NumPy provides functions to perform common linear algebra calculations.



**Demonstrate the Concept**

**Example:**

* Matrix A = [[2, 1], [1, 2]]
* Matrix B = [5, 6]
* Determinant of A = 3
* Inverse of A = [[0.67, -0.33], [-0.33, 0.67]]
* Solution of Ax = B = [1.33, 1.67]
* Eigenvalues of A = 3 and 1
  
**AI/ML Usage:**

* Linear algebra is used in Machine Learning for calculations involving features, weights, matrices, and model training.

```python
# Implement the Concept:
# matrix
matrix_1 = np.array([[2, 1],[1, 2]])
matrix_2 = np.array([5, 6])
# Determinant - matrix_1
determinant = np.linalg.det(matrix_1)

# Inverse - matrix_1
inverse = np.linalg.inv(matrix_1)

# Linear algebra
solution = np.linalg.solve(matrix_1, matrix_2)
eigenvalues, eigenvectors = np.linalg.eig(matrix_1)

print("Determinant:", determinant)
print("Inverse:\n",inverse )
print("Solution to Ax = b:", solution)
print("Eigenvalues:", eigenvalues)
print("Eigenvectors:\n", eigenvectors)
```

## **Best Practices**

* Using arrays with incompatible shapes.
* Confusing argmin and minimum.
* Confusing argmax and maximum.

* Using matrix multiplication when element-wise multiplication is needed.
