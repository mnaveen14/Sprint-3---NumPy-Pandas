## **1. isnull()**



**Understand the Concept**

* isnull() is used to identify missing values in a DataFrame.
* It returns True for missing values and False for non-missing values.
* It can be used to check individual values, columns, or the entire DataFrame.



**Demonstrate the Concept**

**Real-Life Example:**

* An employee dataset may have missing values in Age, Salary, or Department.
* isnull() helps identify which columns and roes contain missing information.

**AI/ML Usage:**

* isnull() is used during data preprocessing to identify missing values before training a Machine Learning model.

```python
# Implement the Concept:
# isnull() - check missing value in each col 
employees.isnull().sum()
```

**Interpretation**

* isnull() identifies missing values in the dataset.
* sum() counts the number of missing values in each column.

## **2. notnull()**



**Understand the Concept**

* notnull() is used to identify non-missing values in a DataFrame.
* It returns True for values that are present and False for missing values.
* It is the opposite of isnull().



**Demonstrate the Concept**

**Real-Life Example:**

* An employee dataset may contain some missing Salary values.
* notnull() can be used to identify employees whose Salary information is available.

**AI/ML Usage:**

* notnull() can be used to select records with available values before performing data analysis or Machine Learning preprocessing.

```python
# Implement the Concept:
# notnull() - check non missing values in each col
employees.notnull().sum()
```

**Interpretation**

* notnull() identifies non-missing values in the dataset.
* sum() counts the available values in each column.

## **3. dropna()**



**Understand the Concept**

* dropna() is used to remove rows or columns containing missing values.
* By default, it removes rows that contain at least one missing value.
* It can also remove columns with missing values using the axis parameter.



**Demonstrate the Concept**

**Real-Life Example:**

* A company wants to train a Machine Learning model using only complete employee records.
* dropna() can remove records that contain missing information.

**AI/ML Usage:**

* dropna() can be used during data preprocessing when the number of missing records is small and removing them does not significantly affect the dataset.

```python
# Implement the Concept:
# dropna() - remove missing values in rows
employees_clean = employees.dropna()

print("Before dropna() : ", employees.shape)
print("After dropna() : ", employees_clean.shape)
```

dropna() removes rows with missing values.

## **4. fillna()**



**Understand the Concept**

* fillna() is used to replace missing values with a specified value or statistical value such as mean, median, or mode.



**Demonstrate the Concept**

**Real-Life Example:**

* A company has missing salary information for some employees.
* Instead of removing those employee records, fillna() can replace the missing values with the average salary.

**AI/ML Usage:**

* fillna() helps retain records by replacing missing values before training a Machine Learning model.

```python
# Implement the Concept:
# fillna() - fill missing salary with mean salary
employees["Salary"] = employees["Salary"].fillna(employees["Salary"].mean())

# check missing values
print(employees["Salary"].isnull().sum())
```

fillna() replaces missing Salary values with the mean Salary.

## **5. interpolate()**



**Understand the Concept**

* interpolate() fills missing values by estimating them from nearby values.
* It is commonly used with numerical or time-series data.



**Demonstrate the Concept**

**Real-Life Example:**

* An employee's performance score is missing between two recorded scores.
* interpolate() can estimate the missing value based on nearby values.

**AI/ML Usage:**

* interpolate() can help fill missing numerical values while preserving the data pattern.

```python
# Implement the Concept:
print("Before interpolation : ",employees["Experience"].isnull().sum())

# interpolate() - fill missing Experience val using interpolation
employees["Experience"] = employees["Experience"].interpolate()

print("After interpolation : ",employees["Experience"].isnull().sum())
```

interpolate() estimates missing Experience values using nearby values.

## **6. Missing Value Handling Strategies**

**Understand the Concept**

* There is no single method that works for every missing value.
* The appropriate strategy depends on:
    * Amount of missing data
    * Data type
    * Data distribution
    * Importance of the column
    * Meaning of the missing value
    * Impact on analysis or Machine Learning


**Demonstrate the Concept**

For the employee dataset, different columns can use different strategies.

**AI/ML Usage:**

* Missing values must be handled before training many Machine Learning models.
* Numerical and categorical features may require different strategies.
* The chosen method should preserve useful information and avoid introducing unnecessary bias.




### **Strategy 1 – Drop Rows**

Use this when only a small number of records have missing values and removing them does not significantly reduce the dataset.



**Real-Life Example:**

* If an employee name is missing, the record may be difficult to identify.
* Such records can be removed when only a few are affected.

```python
print(employees.shape)
```

```python
# Drop Rows - remove rows with missing Name values
employees = employees.dropna(subset=["Name"])
print(employees.shape)
```

### **Strategy 2 – Drop Columns**

Use this when a column contains a very high percentage of missing values and does not have important information.

**Real-Life Example:**

If Previous_Company col is missing for most employees and is not important for the analysis, the column can be removed.

```python
# Previous_Company - col with null value
employees["Previous_Company"] = pd.NA
employees.loc[0:24, "Previous_Company"] = "XYZ Company"  # add some value
employees["Previous_Company"].isnull().sum()
```

```python
# drop col - remove Previous_Company col
employees = employees.drop(columns=["Previous_Company"])

print(employees.shape)
```

### **Strategy 3 – Mean Imputation**

Use the mean for numerical columns when the data distribution is reasonably balanced.

**Real-Life Example:**

* Some employee salary values are missing.
* The average salary can be used to fill the missing values.

```python
# Mean Imputation - fill missing Salary values with mean
employees["Salary"] = employees["Salary"].fillna(employees["Salary"].mean())
print(employees["Salary"].isnull().sum())
```

### **Strategy 4 – Median Imputation**

Use the median when numerical data contains extreme values or has a skewed distribution.

**Real-Life Example:**

* A few employees age have unusually high or low values.
* The median Age provides a more stable replacement for missing values.

```python
# Median Imputation - fill missing Age values with median
employees["Age"] = employees["Age"].fillna(employees["Age"].median())
print(employees["Age"].isnull().sum())
```

### **Strategy 5 – Mode Imputation**

Use the mode for categorical columns.

**Real-Life Example:**

* Some employee department values are missing.
* The most common department can be used to fill those missing values.

```python
# Mode Imputation - fill missing Department values with mode
employees["Department"] =employees["Department"].fillna(employees["Department"].mode()[0])
print(employees["Department"].isnull().sum())
```

### **Strategy 6 – Forward Fill**

Forward fill uses the previous available value to fill a missing value.

**Real-Life Example:**

Employee records have a missing Experience value and we fill with the previous available value.

```python
# Forward Fill - fill missing Experience values using the previous value
employees["Experience"] = employees["Experience"].ffill()
print(employees["Experience"].isnull().sum())
```

### **Strategy 7 – Backward Fill**

Backward fill uses the next available value to fill a missing value.

**Real-Life Example:**

In employee records have a missing performance score can be filled using the next available value.

```python
# Backward Fill - fill missing Performance_Score using the next available value
employees["Performance_Score"] = (employees["Performance_Score"].bfill())
print(employees["Performance_Score"].isnull().sum())
```

```python
# Check missing values after handling
print(employees.isnull().sum())
```

**Interpretation**

* Different columns require different missing value handling methods.
* Mean and median are useful for numerical data.
* Mode is useful for categorical data.
* Forward fill and backward fill can be useful when the row order has meaning.
* Dropping rows or columns should be done carefully to avoid losing important data.

## **Best Practices**

* Check missing values before applying any handling method.
* Choose the strategy based on the data type and amount of missing data.
* Use mean or median for numerical columns when appropriate.
* Use mode for categorical columns.
* Use forward fill or backward fill only when the row order has meaning.
* Avoid dropping large amounts of useful data.
* Verify missing values after handling.

## **Common Mistakes**

* Filling all columns with the same value without considering the data type.
* Using mean when extreme values strongly affect the data.
* Applying forward fill or backward fill to randomly ordered data.
* Dropping too many rows or important columns.
* Filling missing values without checking the data distribution.

## **Advantages**

* Helps create a complete dataset.
* Reduces errors during analysis and Machine Learning.
* Preserves useful records.
* Supports different strategies for different data types.
* Improves data quality.

## **Limitations**

* Incorrect imputation can introduce bias.
* Dropping data may reduce the dataset size.
* Mean and mode may not represent all missing values accurately.
* Forward and backward fill depend on meaningful data order.
